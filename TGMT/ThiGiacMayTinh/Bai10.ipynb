```python
# Import thu vien
from PIL import Image
import matplotlib.pyplot as plt
            
# duong dan anh
my_path = 'C:/Users/admin/OneDrive - Dong Nai Technology University (DNTU)/TGMT/ThuVien/images4.jpg'

# doc anh
im = Image.open(my_path)

# Chuyen doi backend
plt.switch_backend('tkagg')

# Hiển thị ảnh
plt.imshow(im)
plt.title('Click on the image to select points')

# Sử dụng hàm ginput để chọn điểm trên ảnh
points = plt.ginput(5)

print(points)

# Hiển thị
plt.show()
```

```python
# ve lai cac diem toa do bang dau * mau do 
plt.close()

# ve lai
plt.imshow(im)
for point in points:
    x, y = point
    plt.plot(x,y, 'r*')

plt.show()
```
